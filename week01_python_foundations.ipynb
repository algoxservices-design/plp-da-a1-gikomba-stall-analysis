```python
daily_sales = [
    2300, 1950, 2800, 2100, 3600, 5200, 4100,
    2500, 2200, 3100, 2400, 3900, 5600, 4700
]

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg)
# cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140}
}

print("Dataset successfully initialized!")
```

Daily sales analysis
Task 1: Calculate total, average, highest and lowest sales. Identify the corresponding days and weeks.

```python
# Calculate total sales
total_sales = sum(daily_sales)

# Calculate average sales
average_sales = total_sales / len(daily_sales)

# Find highest and lowest sales
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Find their positions
best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

# Identify the corresponding days
best_day = days[best_index]
worst_day = days[worst_index]

# Identify the week numbers
best_week = (best_index // 7) + 1
worst_week = (worst_index // 7) + 1

print("DAILY SALES ANALYSIS")
print("---------------------")

print(f"Total Sales: KES {total_sales:,}")
print(f"Average Daily Sales: KES {average_sales:,.2f}")
print(f"Highest Sales: KES {highest_sales:,}")
print(f"Lowest Sales: KES {lowest_sales:,}")

print(f"\nBest Day: {best_day}, Week {best_week}")
print(f"Worst Day: {worst_day}, Week {worst_week}")
```

Weekly performance and growth
Task 2: Create a function that calculates weekly sales totals and week-on-week growth.

```python
def week_totals(sales):

    week1 = sum(sales[:7])

    week2 = sum(sales[7:])

    return week1, week2


# Call the function and unpack the returned values
week1_total, week2_total = week_totals(daily_sales)


# Calculate percentage growth
growth = ((week2_total - week1_total) / week1_total) * 100


print("WEEKLY PERFORMANCE")
print("-------------------")

print(f"Week 1 Sales: KES {week1_total:,}")
print(f"Week 2 Sales: KES {week2_total:,}")

print(f"Week-on-Week Growth: {growth:.2f}%")
```

Classifying daily sales
Task 3: Create a function that classifies sales into Strong, Normal and Weak categories.

```python
def classify_day(amount, average):

    if amount > average * 1.20:
        return "Strong"

    elif amount < average * 0.80:
        return "Weak"

    else:
        return "Normal"


# Create a dictionary to count categories
sales_categories = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}


# Classify every day's sales
for amount in daily_sales:

    category = classify_day(amount, average_sales)

    sales_categories[category] += 1


# Display results
print("DAILY SALES CLASSIFICATION")
print("--------------------------")

for category, count in sales_categories.items():

    print(f"{category}: {count} days")
```

Weekday vs. weekend performance
Task 4: Compare average sales during weekdays and weekends.

```python
# Separate weekday and weekend sales

weekday_sales = []

weekend_sales = []


for i in range(len(daily_sales)):

    if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]:

        weekday_sales.append(daily_sales[i])

    else:

        weekend_sales.append(daily_sales[i])


# Calculate averages

weekday_average = sum(weekday_sales) / len(weekday_sales)

weekend_average = sum(weekend_sales) / len(weekend_sales)


# Calculate percentage difference

percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100


print("WEEKDAY VS WEEKEND ANALYSIS")
print("---------------------------")

print(f"Weekday Average: KES {weekday_average:,.2f}")

print(f"Weekend Average: KES {weekend_average:,.2f}")

print(f"Weekend Performance Difference: {percentage_difference:.2f}%")
```

Inventory analysis and gross margins
Task 5: Identify products requiring restocking and calculate their gross margin percentages.

```python
# Function to identify products requiring restocking

def products_to_restock(stock):

    restock_list = []

    for product, details in stock.items():

        if details["qty"] <= details["reorder"]:

            restock_list.append(product)

    return restock_list


# Identify products requiring restocking

restock_products = products_to_restock(stock)


print("PRODUCTS REQUIRING RESTOCKING")
print("-----------------------------")

for product in restock_products:

    print(product)


# Calculate gross margins for every product

print("\nPRODUCT GROSS MARGINS")
print("--------------------")

for product, details in stock.items():

    cost = details["cost"]

    price = details["price"]

    margin = ((price - cost) / price) * 100

    print(f"{product}: {margin:.2f}%")
```
